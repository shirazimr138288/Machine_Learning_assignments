```python
import pandas as pd

data = {
    'math':[42,87,15,63,98,54,29,76,11,68],
    'economy':[33,71,58,90,26,47,82,66,5,79],
    'literature':[12,88,45,67,53,30,91,76,18,60],
    'arabic':[7,64,82,26,56,21,93,48,15,77],
    'history':[26,58,82,12,71,38,95,44,17,63],
    'philosophy':[9,55,78,31,67,42,89,50,23,88],
    'logic':[14,61,88,27,73,46,90,33,19,69],
    'psychology':[5,68,81,34,77,49,92,41,26,74],
    'accepted':[1,1,1,0,1,0,1,1,0,0]
}

df = pd.DataFrame(data)
X = df[['math','economy','literature','arabic','history','philosophy','logic','psychology']]
y = df['accepted']
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.4, random_state=58)
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test) 
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```
