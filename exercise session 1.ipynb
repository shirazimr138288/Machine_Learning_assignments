```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
x=np.linspace(0, 2*np.pi, 200)
y=np.sin(x)
plt.plot(x,y)
plt.xlabel("Number")
plt.ylabel("Amplitude")
plt.title("Sin(X) Plot")
plt.grid(True)
plt.show()
```
